```python
import os
import json

NUM_ROUNDS = 24

exp_dir = '/home/ikapetan/Frameworks/Projects-Master/FACT/FACT29/games_descriptions/base/output/all_greedy'
```

```python
should_not_exist = ['<plan>', '<scratchpad>']
# should_not_exist = ['plan', '<plan>', 'scratchpad', '<scratchpad>']
# should_not_exist = ['plan', '<plan>', '</plan>', 'scratchpad', '<scratchpad>', '</scratchpad>']

def contains_leak(public_answer):
    public_answer = public_answer.lower().split()
    for word in should_not_exist:
        if word in public_answer:
            return True
    return False
    
```

```python
def evaluate_game(answers):
    success = answers['finished_rounds'] == NUM_ROUNDS+1
    rounds_leaked = 0
    for round in answers['rounds']:
        pub_ans = round['public_answer']
        leaked = contains_leak(pub_ans)
        if leaked:
            rounds_leaked += 1
    return success, rounds_leaked
```

```python
def evaluate_experiment(exp_dir):
    answers_files = [ os.path.join(exp_dir,filename) for filename in os.listdir(exp_dir) if filename.startswith("history")]
    total_rounds_leaked = []
    total_rounds = []
    successfull_games = []
    for file in answers_files:
        # read json
        answers = json.load(open(file))
        success, rounds_leaked = evaluate_game(answers)
        total_rounds.append(len(answers['rounds']))
        total_rounds_leaked.append(rounds_leaked)
        successfull_games.append(success)
    
    print("Successfull games %d/%d | %.2f%%" % (sum(successfull_games), len(successfull_games), sum(successfull_games)/len(successfull_games)*100))
    print("Leaked games %d/%d | %.2f%%" % (sum([r > 0 for r in total_rounds_leaked]), len(total_rounds_leaked), sum([r > 0 for r in total_rounds_leaked])/len(total_rounds_leaked)*100))
    print("Percentage of rounds leaked %.2f%%" % (sum(total_rounds_leaked)/sum(total_rounds)*100))
  
    successfull_games_percentage = sum(successfull_games)/len(successfull_games)*100
    leaked_games_percentage = sum([r > 0 for r in total_rounds_leaked])/len(total_rounds_leaked)*100
    rounds_leaked_percentage = sum(total_rounds_leaked)/sum(total_rounds)*100
    return successfull_games_percentage, leaked_games_percentage, rounds_leaked_percentage



```

```python
evaluate_experiment(exp_dir)
```
