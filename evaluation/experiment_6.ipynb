# Experiment 6: **Game Statistics**

## Objective
In this experiment, we evaluate some internal statistics of the different provided games.

---

## Methodology
- **Games Evaluated:**  
  - Base  
  - Base Rewritten  
  - Game1  
  - Game2  
  - Game3  

- **Statistics Computed:**  
  - $|\Pi_{acc}|$: Size of the set of acceptable deals
  - $|\Pi_{hard}|$: Size of the set of 6-way acceptable deals
  - Sparsity : Percentage of zero-valued weights in the score functions
  - IoU: Overlap between the score functions as defined in section 4.2.2
---

## Results
The results of this experiment analyze the structural properties of the games and their scoring functions. These findings are presented in **Table 8** of our paper.


```python
import os
import eval_utils as evaluation
import json
import numpy as np
import pandas as pd
from IPython.display import display
import matplotlib.pyplot as plt

raw_path = '../our_games_descriptions'
games = ['base', 'base_rewritten', 'game1', 'game2', 'game3']

results = {}
ISSUES_NUM = 5
AGENTS_NUM = 6

for game in games:
    directory = os.path.join(raw_path, game, 'output', 'original_code', 'gpt4o-mini') # We used gpt4o-mini since the statistics computed are independent of the model used
    agents, role_to_agents, incentive_to_agents = evaluation.load_setup(directory, AGENTS_NUM, num_issues=ISSUES_NUM)
    answers_files = [ os.path.join(directory,filename) for filename in os.listdir(directory) if filename.startswith("history")]

    all_deals = evaluation.get_all_deals(agents)
    feasibility_set = evaluation.compute_feasibility_set(agents, all_deals)
    hard_feasibility_set = evaluation.compute_hard_feasibility_set(agents, all_deals)
    sparsity = evaluation.get_sparsity(agents)
    iou = evaluation.get_iou(agents)

    results[game] = {
        'feasibility_set': len(feasibility_set),
        'hard_feasibility_set': len(hard_feasibility_set),
        'sparsity': f"{sparsity*100:.2f}",
        'iou': f"{iou*100:.2f}"
    }


df = pd.DataFrame(results).T

# Initialize table storage
table_data = []

# Add data to table
for game in games:
    table_data.append([game, results[game]['feasibility_set'], results[game]['hard_feasibility_set'], results[game]['sparsity'], results[game]['iou']])

# Create table
table = pd.DataFrame(table_data, columns=['Game', 'Feasibility Set', 'Hard Feasibility Set', 'Sparsity', 'IoU'])
display(table)
```
